# `Matplotlib` and `Seaborn` are two most popular libraries for data visualization.

```python
import matplotlib.pyplot as plt
import seaborn as sns

year = ["2007","2011","2015","2019","2023","2025"] #year
people = [6.61,7,7.35,7.71,8.05,8.19] #avg people on earth in billion
plt.plot(year,people,marker = 'o', color='red',linestyle='--')
plt.grid()
plt.title("Avg people on earth")
plt.xlabel("Year")
plt.ylabel("People")
plt.show()
```

# **Comparison**

```python
year2 = ["1983","1987","1991","1995","1999","2003","2007"] #year
people2 = [4.70,5,5.33,5.68,6.02,6.34,6.61] #avg people on earth in billion
plt.plot(year2,people2,marker = 'o', color='blue',linestyle='--')
plt.grid()
plt.title("Avg people on earth")
plt.xlabel("Year")
plt.show()
```

```python
plt.plot(year2,people2)
plt.plot(year,people)
plt.grid()
```

# Bar Plot

```python
teams = ["Real Madrid","FCB","Man City","Man United"]
ucl = [15,5,1,3]
plt.bar(teams,ucl)
plt.xlabel("Teams")
plt.ylabel("Total UCL Won")
plt.show()
```

# Histogram

```python
import numpy as np
marks = [40,44,45,38,35,50,47,41,35,30]
plt.hist(marks,bins=6,edgecolor="black")
plt.xlabel("")
plt.ylabel("")
plt.show()
```

# ScatterPlot

```python
x = np.random.rand(10)
y = np.random.rand(10)

plt.scatter(x, y, color='red', alpha=1)
plt.grid()
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Scatter Plot Example")
plt.show()
```

# Pie Chart

```python
labels = ['Apple', 'Banana', 'Orange', 'Grapes']
qty = [30, 20, 40, 10]

plt.pie(qty,labels=labels)
plt.title("Pie Chart Example")
plt.show()
```

# Heat Map

```python
matrix = np.random.rand(5,5)
plt.imshow(matrix, cmap='coolwarm', interpolation='nearest')
plt.colorbar()
plt.title("Heatmap Example")
plt.show()
```
